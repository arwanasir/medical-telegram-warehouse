```python
import os
from dotenv import load_dotenv, find_dotenv

# Automatically find the .env file in parent folders
env_path = find_dotenv()

if not env_path:
    print("❌ Still not found! Please check if the file is named exactly '.env' (with the dot).")
else:
    load_dotenv(env_path)
    print(f"✅ Found .env at: {env_path}")
    
    # Test loading
    api_id = os.getenv('TG_API_ID')
    if api_id:
        print(f"✅ API_ID loaded successfully: {api_id}")
    else:
        print("❌ .env found, but TG_API_ID is missing or empty inside it.")
```

```python
scraper = TelegramScraper()

async def main():
    await scraper.start()
    channels = ['CheMed12', 'lobelia4cosmetics', 'tikvahpharma']
    for channel in channels:
        print(f"Currently scraping: {channel}")
        await scraper.scrape_channel(channel, limit=100)
    await scraper.disconnect()

# Run the async main function
await main()
```
